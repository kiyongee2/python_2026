```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 샘플 월별 매출 데이터
# CSV로 교체 가능: month,sales
months = np.arange(1,13)
sales = np.array([120,135,128,150,160,172,180,190,210,220,230,250])

df = pd.DataFrame({'month': months, 'sales': sales})
print('[월별 매출 데이터]')
print(df)

# 모델 학습
X = df[['month']]
y = df['sales']
model = LinearRegression()
model.fit(X, y)

# 다음 3개월 예측
future = pd.DataFrame({'month':[13,14,15]})
pred = model.predict(future)
forecast = future.copy()
forecast['predicted_sales'] = pred.round(1)

print('\n[다음 3개월 매출 예측]')
print(forecast)

# 시각화
plt.figure(figsize=(10,5))
plt.plot(df['month'], df['sales'], marker='o', label='Actual Sales')
plt.plot(future['month'], pred, marker='x', linestyle='--', label='Forecast')
plt.title('Sales Forecast System')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# 간단 인사이트
avg_growth = (sales[-1]-sales[0])/(len(sales)-1)
print('\n[AI 인사이트]')
print(f'- 월 평균 증가 추정치: {avg_growth:.1f}')
print('- 현재 추세 유지 시 다음 분기 성장 가능성이 높습니다.')
print('- 성수기 대비 재고 및 마케팅 예산을 사전 확보하세요.')
```
