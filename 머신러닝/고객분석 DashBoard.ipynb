```python
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

st.set_page_config(page_title='고객 분석 대시보드', layout='wide')
st.title('📊 고객 분석 대시보드')

# 샘플 데이터 생성
np.random.seed(42)
rows = 200
df = pd.DataFrame({
    'customer_id': range(1, rows+1),
    'age': np.random.randint(20, 60, rows),
    'gender': np.random.choice(['남성','여성'], rows),
    'purchase_count': np.random.randint(1, 20, rows),
    'total_spend': np.random.randint(50000, 1000000, rows),
    'region': np.random.choice(['서울','경기','인천','부산','대구'], rows)
})

df['grade'] = pd.cut(df['total_spend'], bins=[0,200000,500000,1000000], labels=['Bronze','Silver','Gold'])

# 사이드바 필터
st.sidebar.header('필터')
region = st.sidebar.multiselect('지역 선택', options=df['region'].unique(), default=list(df['region'].unique()))
gender = st.sidebar.multiselect('성별 선택', options=df['gender'].unique(), default=list(df['gender'].unique()))
filtered = df[df['region'].isin(region) & df['gender'].isin(gender)]

# KPI
c1,c2,c3,c4 = st.columns(4)
c1.metric('고객 수', len(filtered))
c2.metric('총 매출', f"{filtered['total_spend'].sum():,}원")
c3.metric('평균 구매 횟수', round(filtered['purchase_count'].mean(),1))
c4.metric('평균 객단가', f"{int(filtered['total_spend'].mean()):,}원")

# 차트 1 지역별 고객수
st.subheader('지역별 고객 분포')
fig, ax = plt.subplots(figsize=(8,4))
filtered['region'].value_counts().plot(kind='bar', ax=ax)
ax.set_ylabel('고객 수')
st.pyplot(fig)

# 차트 2 등급 분포
st.subheader('고객 등급 분포')
fig2, ax2 = plt.subplots(figsize=(8,4))
filtered['grade'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=ax2)
ax2.set_ylabel('')
st.pyplot(fig2)

# AI 인사이트(규칙 기반)
st.subheader('🤖 AI 고객 인사이트')
msg=[]
if filtered['total_spend'].mean() > 500000:
    msg.append('핵심 고객 비중이 높습니다. VIP 프로모션을 추천합니다.')
if filtered['purchase_count'].mean() < 8:
    msg.append('재구매 빈도가 낮습니다. 쿠폰/리마케팅 캠페인을 진행해보세요.')
if '서울' in region:
    msg.append('서울 지역 고객 대상 오프라인 이벤트 효과가 기대됩니다.')
if not msg:
    msg.append('고객 지표가 안정적입니다. 현재 전략을 유지하세요.')
for m in msg:
    st.write('- ' + m)

st.subheader('원본 데이터')
st.dataframe(filtered, use_container_width=True)
```
