```python
# 네이버 환율 정보
import requests
from bs4 import BeautifulSoup

# 네이버 환율 정보 페이지 URL
url = "https://finance.naver.com/marketindex/"

# 페이지 요청 및 HTML 파싱
response = requests.get(url)
html = BeautifulSoup(response.text, "html.parser")
print(html)
```

```python
# 환율 정보 제목 추출
title = html.select_one("h2.h_market1").text.strip()
print("환율 정보 제목:", title)
```

```python
# 환율 정보 리스트 추출 - select()
all_li = html.select("div.market1 ul li")
print(all_li)

# 환율 종류 - select_one()
exchange_name = html.select_one("span.blind").text.strip()
print("환율 종류:", exchange_name)

# 환율 가격 - select_one()
exchange_price = html.select_one("span.value").text.strip()
print("환율 가격:", exchange_price)
```

```python
# 환율 종류 및 가격 출력
for li in all_li:
    exchange_name = li.select_one("span.blind").text.strip()
    exchange_price = li.select_one("span.value").text.strip()
    # print(f"{exchange_name}: {exchange_price}")
    print(f"{exchange_name.split()[0]}: {exchange_price}")
```
