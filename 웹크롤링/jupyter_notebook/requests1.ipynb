```python
from bs4 import BeautifulSoup

url = "https://www.seoul.go.kr/main/index.jsp"
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')
print(soup.title)
```

```python
# 메뉴 글자 추출
# find() : 조건에 맞는 첫 번째 요소 반환
first_li = soup.find('li', attrs={'class': 'public'})
print(first_li)
print(first_li.text)
# print(first_li.get_text())
```

```python
# find_all() : 조건에 맞는 모든 요소 반환
div = soup.find('div', attrs={'class': 'm_service'})
all_li = div.find_all('li')
# print(all_li)

for li in all_li:
    print(li.text)
    
print(all_li[1].text) #응답소(민원신고)
```

```python
# select_one() : CSS 선택자로 조건에 맞는 첫 번째 요소 반환
first_li = soup.select_one('div.m_service > ul > li.public')
print(first_li)
print(first_li.text)
```

```python
# select() : CSS 선택자로 요소 반환
li = soup.select('div.m_service > ul > li')
for item in li:
    print(item.text)
```
